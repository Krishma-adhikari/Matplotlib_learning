# Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1,8])
y = np.array([0,450])
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,'o')
```

```python
plt.plot(y)
```

```python
plt.plot(x,y,marker = 'D')
```

```python
x = np.array([2,3,4,5,6])
y = np.array([20,30,56,42,70])
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, 'o--r')
```

```python
plt.plot(y,marker = 'o',mfc = 'b',mec = 'r',ms = 10)
```

```python
plt.plot(x,y,ls = ':')
```

```python
plt.plot(x,y,c = 'hotpink',linestyle = 'solid',linewidth = '20',marker = 'D')
```

## Multiple Lines

```python
x1 = np.array([2,5,3,2])
x2 = np.array([3,7,2,4])
y1 = np.array([30,40,20,45])
y2 = np.array([30,45,23,50])
plt.plot(x1,y1,x2,y2)
plt.show()
```

# Labels

```python
plt.xlabel("A")
plt.ylabel("B")
font1 = {'family':'serif','color':'red','size':10}
plt.title("Krishma",fontdict = font1,loc = 'right')
plt.plot(x,y)
```

```python
plt.grid(axis = 'y')
plt.plot(y)
```

### Matplotlib subplot

```python
plt.subplot(1,2,2)
plt.plot(x,y)
plt.title("Expenditure")

plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("Income")

plt.suptitle("HIII")
plt.show()
```

```python
plt.subplot(2,1,2)
plt.plot(x,y)


plt.subplot(2,1,1)
plt.plot(x,y)

plt.suptitle("My shop")
plt.show()
```

```python
plt.scatter(x,y,c = 'g')
```

```python
colors = np.array([0, 10, 20, 30, 40])
sizes = np.array([2,6,120,14,50])
plt.scatter(x, y, c=colors, cmap='viridis',s = sizes)
plt.colorbar()
plt.show()
```

```python
colors = np.array([0, 10, 20, 30, 40])
sizes = np.array([2,6,120,14,50])
plt.scatter(x, y, c=colors, cmap='viridis',s = sizes,alpha = 0.5)
plt.colorbar()
plt.show()
```

```python
x = np.random.randint(100,size=(100))
y = np.random.randint(100,size=(100))
colors = np.random.randint(100,size=(100))
sizes = 10 * np.random.randint(100,size=(100))
plt.scatter(x, y, c=colors, cmap='viridis',s = sizes,alpha = 0.5)
plt.colorbar()
plt.show()
```

```python
### Matplotlib bars
```

```python
plt.bar(x,y)
plt.show()
```

```python
#Horizontal bars
plt.barh(x,y)
plt.show()
```

```python
#Color
plt.bar(x,y,color = 'pink')
plt.show()
```

```python
#width 
plt.bar(x,y,width = 0.1)
plt.show()
```

```python
#height 
plt.barh(x,y,height = 0.3)
plt.show()
```
