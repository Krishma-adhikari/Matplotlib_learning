```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.random.randint(10,size=(7))
plt.pie(x)
plt.show()
```

```python
#Labels
mylabel = ["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
plt.pie(x,labels = mylabel)
plt.show()
```

```python
# Start Angle
font1 = {'family': 'serif'}
plt.pie(x,labels = mylabel,startangle = 45)
plt.title("Arrival rate of people in funpark-Days",color = 'r',size = 20)
plt.show()
```

```python
#Explode
myexplode = [0,0.3,0,0,0,0.3,0]
plt.pie(x,explode = myexplode)
plt.show()
```

```python
#Shadows
plt.pie(x,explode = myexplode, shadow = True)
plt.show()
```

```python
#colors
mycolor = ['c','k','r','b','g','#4CFE03',"#1C4F0D"]
plt.pie(x,labels = mylabel,colors = mycolor)
plt.show()
```

```python
#Legend 
plt.pie(x,labels = mylabel)
plt.legend(title = "Day : ",loc = 'lower right')
plt.show()
```
